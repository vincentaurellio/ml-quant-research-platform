```python
import pandas as pd

data_raw = pd.read_parquet("../data/raw/market_data.parquet")

print(data_raw.head())
print(data_raw.dtypes)
print(data_raw.groupby("Ticker").size())
print(data_raw.isna().sum())
```

```python
data = pd.read_parquet("../data/processed/model_dataset.parquet")

target = "target_relative_return_5"

print("Target availability:")
print(data[target].notna().value_counts())

print("\nTarget summary:")
print(data[target].describe(percentiles=[0.01, 0.25, 0.5, 0.75, 0.99]))

print("\nValid target observations per date:")
print(
    data.dropna(subset=[target])
    .groupby("Date")["Ticker"]
    .nunique()
    .describe()
)

print("\nMean relative target by date:")
print(data.groupby("Date")[target].mean().dropna().describe())
```
